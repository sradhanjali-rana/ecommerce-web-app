```python
# Create conversion target

session_conversion = website_sessions.copy()

session_conversion["converted"] = (
    session_conversion["website_session_id"]
    .isin(orders["website_session_id"])
    .astype(int)
)

print(session_conversion["converted"].value_counts())
```

```python
# Select features for prediction

features = [
    "is_repeat_session",
    "utm_source",
    "utm_campaign",
    "utm_content",
    "device_type",
    "http_referer"
]

X = session_conversion[features]
y = session_conversion["converted"]

print("Features:")
print(X.columns.tolist())

print("\nTarget:")
print(y.name)

print("\nShape:")
print("X:", X.shape)
print("y:", y.shape)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Categorical and numerical features
categorical_features = [
    "utm_source",
    "utm_campaign",
    "utm_content",
    "device_type",
    "http_referer"
]

numerical_features = [
    "is_repeat_session"
]

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features
        ),
        (
            "num",
            "passthrough",
            numerical_features
        )
    ]
)

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

# Create Logistic Regression pipeline
logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            LogisticRegression(
                max_iter=1000,
                class_weight="balanced"
            )
        )
    ]
)

# Train the model
logistic_model.fit(X_train, y_train)

print("Logistic Regression model trained successfully!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

# Predictions
y_pred = logistic_model.predict(X_test)

# Probability predictions
y_prob = logistic_model.predict_proba(X_test)[:, 1]

# Evaluation metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_prob)

print("Model Evaluation")
print("-----------------")
print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1 Score : {f1:.4f}")
print(f"ROC-AUC  : {roc_auc:.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier

# Random Forest model
random_forest_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            RandomForestClassifier(
                n_estimators=100,
                random_state=42,
                class_weight="balanced",
                n_jobs=-1
            )
        )
    ]
)

# Train Random Forest
random_forest_model.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

```python
# Random Forest predictions

rf_pred = random_forest_model.predict(X_test)
rf_prob = random_forest_model.predict_proba(X_test)[:, 1]

# Evaluation metrics

rf_accuracy = accuracy_score(y_test, rf_pred)
rf_precision = precision_score(y_test, rf_pred)
rf_recall = recall_score(y_test, rf_pred)
rf_f1 = f1_score(y_test, rf_pred)
rf_roc_auc = roc_auc_score(y_test, rf_prob)

print("Random Forest Evaluation")
print("------------------------")
print(f"Accuracy : {rf_accuracy:.4f}")
print(f"Precision: {rf_precision:.4f}")
print(f"Recall   : {rf_recall:.4f}")
print(f"F1 Score : {rf_f1:.4f}")
print(f"ROC-AUC  : {rf_roc_auc:.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, rf_pred))
```

```python
def predict_conversion(
    is_repeat_session,
    utm_source,
    utm_campaign,
    utm_content,
    device_type,
    http_referer
):
    
    new_session = pd.DataFrame({
        "is_repeat_session": [is_repeat_session],
        "utm_source": [utm_source],
        "utm_campaign": [utm_campaign],
        "utm_content": [utm_content],
        "device_type": [device_type],
        "http_referer": [http_referer]
    })

    prediction = logistic_model.predict(new_session)[0]
    
    probability = logistic_model.predict_proba(
        new_session
    )[0][1]

    return prediction, probability


print("Prediction function created successfully!")
```

```python
prediction, probability = predict_conversion(
    is_repeat_session=1,
    utm_source="gsearch",
    utm_campaign="nonbrand",
    utm_content="brand",
    device_type="desktop",
    http_referer="https://www.google.com/"
)

print("Prediction:", prediction)
print(f"Conversion Probability: {probability:.2%}")
```
